# Feature Engineering & Fraud Detection with Decision Trees


```python
# Import required libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns
```

## Step 1: Create Synthetic Fraud Dataset

```python
# Synthetic dataset creation
np.random.seed(42)
n_samples = 1000
data = pd.DataFrame({
    'TransactionID': np.arange(1, n_samples + 1),
    'Amount': np.random.exponential(scale=100, size=n_samples),
    'Type': np.random.choice(['debit', 'credit'], size=n_samples),
    'IsFraud': np.random.choice([0, 1], size=n_samples, p=[0.95, 0.05])
})
data.head()
```

## Step 2: Preprocessing & Feature Engineering

```python
# Encode categorical features
le = LabelEncoder()
data['TypeEncoded'] = le.fit_transform(data['Type'])
# Create log amount feature
data['LogAmount'] = np.log1p(data['Amount'])
features = ['Amount', 'TypeEncoded', 'LogAmount']
X = data[features]
y = data['IsFraud']
```

## Step 3: Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

## Step 4: Hyperparameter Tuning using GridSearchCV

```python
param_grid = {
    'max_depth': [3, 5, 10, None],
    'min_samples_split': [2, 5, 10],
    'criterion': ['gini', 'entropy']
}
dt = DecisionTreeClassifier(random_state=42)
grid = GridSearchCV(dt, param_grid, cv=5, scoring='f1', n_jobs=-1)
grid.fit(X_train, y_train)
print("Best Parameters:", grid.best_params_)
```

## Step 5: Evaluation

```python
y_pred = grid.predict(X_test)
print(classification_report(y_test, y_pred))
```
